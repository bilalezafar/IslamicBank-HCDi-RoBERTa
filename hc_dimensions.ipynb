```python
# ============================================================
# Mount Google Drive, read HC BERT dataset,
# and inspect token sizes for selected transformer models
# ============================================================

# -----------------------------
# 1. Install required libraries
# -----------------------------
!pip -q install transformers sentencepiece accelerate

# -----------------------------
# 2. Mount Google Drive
# -----------------------------
from google.colab import drive
drive.mount("/content/drive")

# -----------------------------
# 3. Import libraries
# -----------------------------
import pandas as pd
import numpy as np
from transformers import AutoTokenizer
import matplotlib.pyplot as plt

# -----------------------------
# 4. Read dataset from Google Drive
# -----------------------------
file_path = "/content/drive/MyDrive/hc_dimensions/hc_sentences_bert.csv"

df = pd.read_csv(file_path, encoding="utf-8-sig", low_memory=False)

print("Rows and columns:", df.shape)
print("Columns:", df.columns.tolist())

# Check required columns
required_cols = ["sentence", "classification"]
missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

# Clean text columns
df["sentence"] = df["sentence"].astype(str).str.strip()
df["classification"] = df["classification"].astype(str).str.strip()

# Remove blank sentences
df = df[df["sentence"].notna()].copy()
df = df[df["sentence"] != ""].copy()

print("\nRows after removing blank sentences:", len(df))

print("\nClass distribution:")
print(df["classification"].value_counts())

# -----------------------------
# 5. Basic sentence length inspection
# -----------------------------
df["word_count"] = df["sentence"].apply(lambda x: len(str(x).split()))
df["char_len"] = df["sentence"].str.len()

print("\nWord count summary:")
display(df["word_count"].describe())

print("\nCharacter length summary:")
display(df["char_len"].describe())

# -----------------------------
# 6. Models
# -----------------------------
models_to_check = {
    "BERT base": "google-bert/bert-base-uncased",
    "RoBERTa base": "FacebookAI/roberta-base",
    "DeBERTa V3 base": "microsoft/deberta-v3-base",
    "FinBERT pretrain": "yiyanghkust/finbert-pretrain"
}

# -----------------------------
# 7. Function to calculate token lengths
# -----------------------------
def calculate_token_lengths(sentences, model_name, batch_size=512):
    tokenizer = AutoTokenizer.from_pretrained(model_name)

    token_lengths = []

    for i in range(0, len(sentences), batch_size):
        batch = sentences[i:i + batch_size]

        encoded = tokenizer(
            batch,
            add_special_tokens=True,
            truncation=False,
            padding=False
        )

        token_lengths.extend([len(ids) for ids in encoded["input_ids"]])

    return token_lengths, tokenizer.model_max_length

# -----------------------------
# 8. Token length inspection
# -----------------------------
sentences = df["sentence"].tolist()

summary_rows = []
token_length_dfs = {}

for model_label, model_name in models_to_check.items():
    print(f"\nChecking tokenizer: {model_label} | {model_name}")

    try:
        token_lengths, model_max_length = calculate_token_lengths(
            sentences=sentences,
            model_name=model_name
        )

        temp = pd.DataFrame({
            "sentence": sentences,
            "classification": df["classification"].tolist(),
            "token_length": token_lengths
        })

        token_length_dfs[model_label] = temp

        summary_rows.append({
            "model_label": model_label,
            "model_name": model_name,
            "model_max_length": model_max_length,
            "mean_tokens": round(np.mean(token_lengths), 2),
            "median_tokens": round(np.median(token_lengths), 2),
            "p75_tokens": round(np.percentile(token_lengths, 75), 2),
            "p90_tokens": round(np.percentile(token_lengths, 90), 2),
            "p95_tokens": round(np.percentile(token_lengths, 95), 2),
            "p99_tokens": round(np.percentile(token_lengths, 99), 2),
            "max_tokens": int(np.max(token_lengths)),
            "sentences_over_64": int(sum(x > 64 for x in token_lengths)),
            "sentences_over_128": int(sum(x > 128 for x in token_lengths)),
            "sentences_over_256": int(sum(x > 256 for x in token_lengths)),
            "sentences_over_512": int(sum(x > 512 for x in token_lengths)),
            "pct_over_64": round(sum(x > 64 for x in token_lengths) / len(token_lengths) * 100, 2),
            "pct_over_128": round(sum(x > 128 for x in token_lengths) / len(token_lengths) * 100, 2),
            "pct_over_256": round(sum(x > 256 for x in token_lengths) / len(token_lengths) * 100, 2),
            "pct_over_512": round(sum(x > 512 for x in token_lengths) / len(token_lengths) * 100, 2)
        })

    except Exception as e:
        print(f"Could not process {model_label}: {e}")

token_summary = pd.DataFrame(summary_rows)

print("\nToken length summary:")
display(token_summary)

# -----------------------------
# 9. Save token summary to Google Drive
# -----------------------------
summary_output_path = "/content/drive/MyDrive/hc_dimensions/hc_token_length_summary_selected_models.csv"

token_summary.to_csv(
    summary_output_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nSaved token summary to:")
print(summary_output_path)

# -----------------------------
# 10. Plot token length distribution
# -----------------------------
for model_label, temp in token_length_dfs.items():
    plt.figure(figsize=(10, 5))
    plt.hist(temp["token_length"], bins=80)
    plt.xlabel("Token length")
    plt.ylabel("Number of sentences")
    plt.title(f"Token Length Distribution: {model_label}")
    plt.show()

# -----------------------------
# 11. Inspect longest sentences by each tokenizer
# -----------------------------
for model_label, temp in token_length_dfs.items():
    print(f"\nLongest sentences based on {model_label} tokenizer:")
    longest = temp.sort_values("token_length", ascending=False).head(20)
    display(longest[["classification", "token_length", "sentence"]])

# -----------------------------
# 12. Suggested max_length decision table
# -----------------------------
decision_rows = []

for _, row in token_summary.iterrows():
    if row["pct_over_128"] <= 5:
        suggested_max_length = 128
    elif row["pct_over_256"] <= 5:
        suggested_max_length = 256
    else:
        suggested_max_length = 512

    decision_rows.append({
        "model_label": row["model_label"],
        "pct_over_128": row["pct_over_128"],
        "pct_over_256": row["pct_over_256"],
        "suggested_max_length": suggested_max_length
    })

decision_table = pd.DataFrame(decision_rows)

print("\nSuggested max_length by model:")
display(decision_table)

decision_output_path = "/content/drive/MyDrive/hc_dimensions/hc_token_max_length_decision.csv"

decision_table.to_csv(
    decision_output_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nSaved max_length decision table to:")
print(decision_output_path)
```

```python
# ============================================================
# Human Capital Disclosure Classification in Islamic Banking
# Full Colab Training Code
#
# Models:
# 1. BERT base
# 2. RoBERTa base
# 3. DeBERTa V3 base
# 4. FinBERT pretrain
#
# Input file:
# /content/drive/MyDrive/hc_dimensions/hc_sentences_bert.csv
#
# Required columns:
# sentence
# classification
#
# Outputs folder:
# /content/drive/MyDrive/hc_dimensions/outputs
#
# Saves:
# 1. CSV results
# 2. SVG figures
# 3. Best model only
#
# Label structure:
# 0 = Not related
# 1 = Diversity, Equity, and Inclusion
# 2 = Compensation and Benefits
# 3 = Labour Relations and Organisational Culture
# 4 = Health, Safety, and Well-being
# 5 = Workforce Demographics
# 6 = Training, Development, and Human Capital Investment
# ============================================================


# ============================================================
# 1. Install required libraries
# ============================================================

!pip -q install transformers datasets accelerate evaluate sentencepiece scikit-learn matplotlib


# ============================================================
# 2. Mount Google Drive
# ============================================================

from google.colab import drive
drive.mount("/content/drive")


# ============================================================
# 3. Imports
# ============================================================

import os
import re
import gc
import json
import time
import math
import shutil
import inspect
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
from torch import nn

from datasets import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding,
    set_seed
)

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)
from sklearn.utils.class_weight import compute_class_weight

warnings.filterwarnings("ignore")


# ============================================================
# 4. Global settings
# ============================================================

RANDOM_STATE = 42
set_seed(RANDOM_STATE)

BASE_DIR = "/content/drive/MyDrive/hc_dimensions"
DATA_FILE = os.path.join(BASE_DIR, "hc_sentences_bert.csv")

OUTPUT_DIR = os.path.join(BASE_DIR, "outputs")
BEST_MODEL_DIR = os.path.join(OUTPUT_DIR, "best_model")
TEMP_RUNS_DIR = "/content/hc_temp_model_runs"

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(TEMP_RUNS_DIR, exist_ok=True)

MAX_LENGTH = 128
NUM_EPOCHS = 5
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.10
USE_CLASS_WEIGHTS = True

# Important:
# Keep fp16 off to avoid dtype problems with DeBERTa on Tesla T4.
USE_FP16 = False
USE_BF16 = False

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()


# ============================================================
# 5. Model list
# ============================================================

MODEL_SPECS = {
    "bert_base": {
        "label": "BERT base",
        "model_name": "google-bert/bert-base-uncased",
        "train_batch_size": 16,
        "eval_batch_size": 32,
        "gradient_accumulation_steps": 1
    },
    "roberta_base": {
        "label": "RoBERTa base",
        "model_name": "FacebookAI/roberta-base",
        "train_batch_size": 16,
        "eval_batch_size": 32,
        "gradient_accumulation_steps": 1
    },
    "deberta_v3_base": {
        "label": "DeBERTa V3 base",
        "model_name": "microsoft/deberta-v3-base",
        "train_batch_size": 4,
        "eval_batch_size": 16,
        "gradient_accumulation_steps": 4
    },
    "finbert_pretrain": {
        "label": "FinBERT pretrain",
        "model_name": "yiyanghkust/finbert-pretrain",
        "train_batch_size": 16,
        "eval_batch_size": 32,
        "gradient_accumulation_steps": 1
    }
}


# ============================================================
# 6. Load and prepare dataset
# ============================================================

df = pd.read_csv(DATA_FILE, encoding="utf-8-sig", low_memory=False)

print("Dataset loaded:", df.shape)
print("Columns:", df.columns.tolist())

required_cols = ["sentence", "classification"]
missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

df = df[["sentence", "classification"]].copy()

df["sentence"] = df["sentence"].astype(str).str.strip()
df["classification"] = df["classification"].astype(str).str.strip()

df = df[df["sentence"].notna()].copy()
df = df[df["sentence"] != ""].copy()
df = df[df["classification"].notna()].copy()
df = df[df["classification"] != ""].copy()

print("\nFinal dataset size:", df.shape)

print("\nClass distribution:")
print(df["classification"].value_counts())


# ============================================================
# 7. Label mapping
# ============================================================
# 0 = Not related
# 1 onward = HC dimensions

preferred_label_order = [
    "Not related",
    "Diversity, Equity, and Inclusion",
    "Compensation and Benefits",
    "Labour Relations and Organisational Culture",
    "Health, Safety, and Well-being",
    "Workforce Demographics",
    "Training, Development, and Human Capital Investment"
]

actual_labels = df["classification"].unique().tolist()

labels = [label for label in preferred_label_order if label in actual_labels]
extra_labels = sorted([label for label in actual_labels if label not in labels])
labels = labels + extra_labels

label2id = {label: i for i, label in enumerate(labels)}
id2label = {i: label for label, i in label2id.items()}

df["label"] = df["classification"].map(label2id)

if df["label"].isna().sum() > 0:
    raise ValueError("Some labels could not be mapped.")

df["label"] = df["label"].astype(int)

num_labels = len(labels)

label_mapping_df = pd.DataFrame({
    "model_label_id": list(range(num_labels)),
    "classification": [id2label[i] for i in range(num_labels)]
})

label_mapping_df.to_csv(
    os.path.join(OUTPUT_DIR, "label_mapping.csv"),
    index=False,
    encoding="utf-8-sig"
)

with open(os.path.join(OUTPUT_DIR, "label_mapping.json"), "w", encoding="utf-8") as f:
    json.dump(
        {
            "label2id": label2id,
            "id2label": {str(k): v for k, v in id2label.items()}
        },
        f,
        indent=2,
        ensure_ascii=False
    )

print("\nLabel mapping:")
display(label_mapping_df)


# ============================================================
# 8. Train, validation, and test split
# ============================================================

train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=df["label"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=RANDOM_STATE,
    stratify=temp_df["label"]
)

print("\nSplit sizes:")
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

print("\nTrain class distribution:")
print(train_df["classification"].value_counts())

print("\nValidation class distribution:")
print(val_df["classification"].value_counts())

print("\nTest class distribution:")
print(test_df["classification"].value_counts())

train_df.to_csv(
    os.path.join(OUTPUT_DIR, "train_split.csv"),
    index=False,
    encoding="utf-8-sig"
)

val_df.to_csv(
    os.path.join(OUTPUT_DIR, "validation_split.csv"),
    index=False,
    encoding="utf-8-sig"
)

test_df.to_csv(
    os.path.join(OUTPUT_DIR, "test_split.csv"),
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 9. Class weights
# ============================================================

classes_array = np.arange(num_labels)

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=classes_array,
    y=train_df["label"].values
)

class_weights_tensor = torch.tensor(class_weights, dtype=torch.float32)

class_weights_df = pd.DataFrame({
    "model_label_id": classes_array,
    "classification": [id2label[i] for i in classes_array],
    "class_weight": class_weights
})

class_weights_df.to_csv(
    os.path.join(OUTPUT_DIR, "class_weights.csv"),
    index=False,
    encoding="utf-8-sig"
)

print("\nClass weights:")
display(class_weights_df)


# ============================================================
# 10. Helper functions
# ============================================================

def clean_key(text):
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9]+", "_", text)
    text = re.sub(r"_+", "_", text).strip("_")
    return text


def compute_metrics(eval_pred):
    logits, labels_true = eval_pred
    preds = np.argmax(logits, axis=1)

    acc = accuracy_score(labels_true, preds)

    macro_precision, macro_recall, macro_f1, _ = precision_recall_fscore_support(
        labels_true,
        preds,
        average="macro",
        zero_division=0
    )

    weighted_precision, weighted_recall, weighted_f1, _ = precision_recall_fscore_support(
        labels_true,
        preds,
        average="weighted",
        zero_division=0
    )

    return {
        "accuracy": acc,
        "macro_precision": macro_precision,
        "macro_recall": macro_recall,
        "macro_f1": macro_f1,
        "weighted_precision": weighted_precision,
        "weighted_recall": weighted_recall,
        "weighted_f1": weighted_f1
    }


class WeightedTrainer(Trainer):
    def __init__(self, class_weights=None, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.class_weights = class_weights

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels_tensor = inputs.get("labels")

        model_inputs = {
            k: v for k, v in inputs.items()
            if k != "labels"
        }

        outputs = model(**model_inputs)
        logits = outputs.get("logits")

        if self.class_weights is not None:
            weights = self.class_weights.to(
                device=logits.device,
                dtype=logits.dtype
            )
            loss_fct = nn.CrossEntropyLoss(weight=weights)
        else:
            loss_fct = nn.CrossEntropyLoss()

        loss = loss_fct(
            logits.view(-1, model.config.num_labels),
            labels_tensor.view(-1)
        )

        return (loss, outputs) if return_outputs else loss


def make_training_args(**kwargs):
    sig = inspect.signature(TrainingArguments.__init__).parameters

    if "eval_strategy" in sig:
        kwargs["eval_strategy"] = "epoch"
    else:
        kwargs["evaluation_strategy"] = "epoch"

    return TrainingArguments(**kwargs)


def make_trainer(
    trainer_class,
    model,
    training_args,
    train_dataset,
    val_dataset,
    tokenizer,
    data_collator,
    class_weights_tensor=None
):
    trainer_kwargs = {
        "model": model,
        "args": training_args,
        "train_dataset": train_dataset,
        "eval_dataset": val_dataset,
        "data_collator": data_collator,
        "compute_metrics": compute_metrics
    }

    trainer_signature = inspect.signature(Trainer.__init__).parameters

    if "processing_class" in trainer_signature:
        trainer_kwargs["processing_class"] = tokenizer
    elif "tokenizer" in trainer_signature:
        trainer_kwargs["tokenizer"] = tokenizer

    if trainer_class == WeightedTrainer:
        trainer_kwargs["class_weights"] = class_weights_tensor

    return trainer_class(**trainer_kwargs)


def make_dataset(input_df, tokenizer):
    temp_df = input_df[["sentence", "label"]].copy()
    dataset = Dataset.from_pandas(temp_df, preserve_index=False)

    def tokenize_batch(batch):
        return tokenizer(
            batch["sentence"],
            truncation=True,
            max_length=MAX_LENGTH,
            padding=False
        )

    dataset = dataset.map(
        tokenize_batch,
        batched=True,
        remove_columns=["sentence"]
    )

    dataset = dataset.rename_column("label", "labels")
    dataset.set_format("torch")

    return dataset


def calculate_warmup_steps(train_size, batch_size, grad_accum_steps, epochs, warmup_ratio):
    steps_per_epoch = math.ceil(train_size / (batch_size * grad_accum_steps))
    total_steps = steps_per_epoch * epochs
    warmup_steps = int(total_steps * warmup_ratio)
    return warmup_steps, total_steps


def save_confusion_matrix_fig(cm, title, labels, save_path, normalize=False):
    fig, ax = plt.subplots(figsize=(10, 8))

    if normalize:
        row_sums = cm.sum(axis=1, keepdims=True)
        cm_to_plot = cm.astype("float") / row_sums
        cm_to_plot = np.nan_to_num(cm_to_plot)
    else:
        cm_to_plot = cm

    im = ax.imshow(cm_to_plot, interpolation="nearest")
    ax.figure.colorbar(im, ax=ax)

    ax.set(
        xticks=np.arange(len(labels)),
        yticks=np.arange(len(labels)),
        xticklabels=labels,
        yticklabels=labels,
        ylabel="True label",
        xlabel="Predicted label",
        title=title
    )

    plt.setp(
        ax.get_xticklabels(),
        rotation=45,
        ha="right",
        rotation_mode="anchor"
    )

    fmt = ".2f" if normalize else "d"

    for i in range(cm_to_plot.shape[0]):
        for j in range(cm_to_plot.shape[1]):
            ax.text(
                j,
                i,
                format(cm_to_plot[i, j], fmt),
                ha="center",
                va="center",
                fontsize=8
            )

    fig.tight_layout()
    plt.savefig(save_path, format="svg", bbox_inches="tight")
    plt.show()
    plt.close()


def save_metric_bar_chart(df_plot, metric_col, title, save_path):
    fig, ax = plt.subplots(figsize=(10, 6))

    x = np.arange(len(df_plot))
    ax.bar(x, df_plot[metric_col])

    ax.set_xticks(x)
    ax.set_xticklabels(df_plot["model_label"], rotation=35, ha="right")
    ax.set_ylabel(metric_col)
    ax.set_title(title)

    for i, v in enumerate(df_plot[metric_col]):
        ax.text(i, v, f"{v:.3f}", ha="center", va="bottom")

    fig.tight_layout()
    plt.savefig(save_path, format="svg", bbox_inches="tight")
    plt.show()
    plt.close()


# ============================================================
# 11. Main training loop
# ============================================================

all_metrics = []
best_val_macro_f1 = -1
best_model_info = None

for model_key, spec in MODEL_SPECS.items():

    print("\n" + "=" * 80)
    print(f"Training model: {spec['label']} | {spec['model_name']}")
    print("=" * 80)

    start_time = time.time()

    model_output_dir = os.path.join(TEMP_RUNS_DIR, model_key)

    if os.path.exists(model_output_dir):
        shutil.rmtree(model_output_dir)

    os.makedirs(model_output_dir, exist_ok=True)

    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    tokenizer = AutoTokenizer.from_pretrained(spec["model_name"])

    train_dataset = make_dataset(train_df, tokenizer)
    val_dataset = make_dataset(val_df, tokenizer)
    test_dataset = make_dataset(test_df, tokenizer)

    data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

    model = AutoModelForSequenceClassification.from_pretrained(
        spec["model_name"],
        num_labels=num_labels,
        id2label=id2label,
        label2id=label2id,
        ignore_mismatched_sizes=True,
        torch_dtype=torch.float32
    )

    model.config.id2label = id2label
    model.config.label2id = label2id

    # Critical dtype fix for DeBERTa and safe for all models
    model = model.float()

    warmup_steps, total_steps = calculate_warmup_steps(
        train_size=len(train_dataset),
        batch_size=spec["train_batch_size"],
        grad_accum_steps=spec["gradient_accumulation_steps"],
        epochs=NUM_EPOCHS,
        warmup_ratio=WARMUP_RATIO
    )

    print(f"Estimated total training steps: {total_steps}")
    print(f"Warmup steps: {warmup_steps}")
    print("fp16 enabled:", USE_FP16)
    print("bf16 enabled:", USE_BF16)
    print("Model dtype check:", next(model.parameters()).dtype)

    training_args = make_training_args(
        output_dir=model_output_dir,
        learning_rate=LEARNING_RATE,
        per_device_train_batch_size=spec["train_batch_size"],
        per_device_eval_batch_size=spec["eval_batch_size"],
        gradient_accumulation_steps=spec["gradient_accumulation_steps"],
        num_train_epochs=NUM_EPOCHS,
        weight_decay=WEIGHT_DECAY,
        warmup_steps=warmup_steps,
        logging_strategy="steps",
        logging_steps=50,

        # No intermediate checkpoint saving.
        # This avoids checkpoint reload dtype issues.
        save_strategy="no",
        load_best_model_at_end=False,

        fp16=USE_FP16,
        bf16=USE_BF16,
        optim="adamw_torch",
        report_to="none",
        seed=RANDOM_STATE,
        data_seed=RANDOM_STATE
    )

    trainer_class = WeightedTrainer if USE_CLASS_WEIGHTS else Trainer

    trainer = make_trainer(
        trainer_class=trainer_class,
        model=model,
        training_args=training_args,
        train_dataset=train_dataset,
        val_dataset=val_dataset,
        tokenizer=tokenizer,
        data_collator=data_collator,
        class_weights_tensor=class_weights_tensor if USE_CLASS_WEIGHTS else None
    )

    train_result = trainer.train()

    train_runtime_minutes = round((time.time() - start_time) / 60, 2)

    # --------------------------------------------------------
    # Validation evaluation
    # --------------------------------------------------------

    val_metrics = trainer.evaluate(eval_dataset=val_dataset)

    val_metrics_clean = {
        k.replace("eval_", "val_"): v
        for k, v in val_metrics.items()
    }

    # --------------------------------------------------------
    # Test prediction
    # --------------------------------------------------------

    test_output = trainer.predict(test_dataset)

    test_logits = test_output.predictions
    test_probs = torch.softmax(torch.tensor(test_logits), dim=1).numpy()
    test_preds = np.argmax(test_logits, axis=1)
    test_true = test_df["label"].values

    test_metrics = compute_metrics((test_logits, test_true))

    test_metrics_clean = {
        f"test_{k}": v
        for k, v in test_metrics.items()
    }

    # --------------------------------------------------------
    # Save classification report
    # --------------------------------------------------------

    report_dict = classification_report(
        test_true,
        test_preds,
        target_names=labels,
        output_dict=True,
        zero_division=0
    )

    report_df = pd.DataFrame(report_dict).transpose().reset_index()
    report_df = report_df.rename(columns={"index": "classification"})

    report_path = os.path.join(
        OUTPUT_DIR,
        f"classification_report_{model_key}.csv"
    )

    report_df.to_csv(report_path, index=False, encoding="utf-8-sig")

    # --------------------------------------------------------
    # Save confusion matrix
    # --------------------------------------------------------

    cm = confusion_matrix(
        test_true,
        test_preds,
        labels=list(range(num_labels))
    )

    cm_df = pd.DataFrame(
        cm,
        index=[f"true_{label}" for label in labels],
        columns=[f"pred_{label}" for label in labels]
    )

    cm_path = os.path.join(
        OUTPUT_DIR,
        f"confusion_matrix_{model_key}.csv"
    )

    cm_df.to_csv(cm_path, encoding="utf-8-sig")

    cm_fig_path = os.path.join(
        OUTPUT_DIR,
        f"confusion_matrix_{model_key}.svg"
    )

    save_confusion_matrix_fig(
        cm=cm,
        title=f"Confusion Matrix: {spec['label']}",
        labels=labels,
        save_path=cm_fig_path,
        normalize=False
    )

    cm_norm_fig_path = os.path.join(
        OUTPUT_DIR,
        f"confusion_matrix_normalized_{model_key}.svg"
    )

    save_confusion_matrix_fig(
        cm=cm,
        title=f"Normalized Confusion Matrix: {spec['label']}",
        labels=labels,
        save_path=cm_norm_fig_path,
        normalize=True
    )

    # --------------------------------------------------------
    # Save test predictions
    # --------------------------------------------------------

    pred_df = test_df[["sentence", "classification", "label"]].copy()

    pred_df = pred_df.rename(columns={
        "classification": "true_classification",
        "label": "true_label_id"
    })

    pred_df["predicted_label_id"] = test_preds
    pred_df["predicted_classification"] = [
        id2label[int(i)] for i in test_preds
    ]

    pred_df["correct"] = (
        pred_df["true_label_id"] == pred_df["predicted_label_id"]
    )

    for i, label in id2label.items():
        safe_label = clean_key(label)
        pred_df[f"prob_{i}_{safe_label}"] = test_probs[:, i]

    pred_path = os.path.join(
        OUTPUT_DIR,
        f"test_predictions_{model_key}.csv"
    )

    pred_df.to_csv(pred_path, index=False, encoding="utf-8-sig")

    # --------------------------------------------------------
    # Save trainer log history
    # --------------------------------------------------------

    log_history_df = pd.DataFrame(trainer.state.log_history)

    log_path = os.path.join(
        OUTPUT_DIR,
        f"trainer_log_history_{model_key}.csv"
    )

    log_history_df.to_csv(log_path, index=False, encoding="utf-8-sig")

    # --------------------------------------------------------
    # Save one-row model metrics
    # --------------------------------------------------------

    row = {
        "model_key": model_key,
        "model_label": spec["label"],
        "model_name": spec["model_name"],
        "max_length": MAX_LENGTH,
        "num_epochs_setting": NUM_EPOCHS,
        "learning_rate": LEARNING_RATE,
        "weight_decay": WEIGHT_DECAY,
        "warmup_steps": warmup_steps,
        "total_steps_estimated": total_steps,
        "train_batch_size": spec["train_batch_size"],
        "eval_batch_size": spec["eval_batch_size"],
        "gradient_accumulation_steps": spec["gradient_accumulation_steps"],
        "use_class_weights": USE_CLASS_WEIGHTS,
        "fp16": USE_FP16,
        "bf16": USE_BF16,
        "train_runtime_minutes": train_runtime_minutes,
        "train_loss": train_result.training_loss,
        **val_metrics_clean,
        **test_metrics_clean
    }

    all_metrics.append(row)

    one_metric_df = pd.DataFrame([row])

    one_metric_path = os.path.join(
        OUTPUT_DIR,
        f"metrics_{model_key}.csv"
    )

    one_metric_df.to_csv(one_metric_path, index=False, encoding="utf-8-sig")

    print("\nValidation metrics:")
    for k, v in val_metrics_clean.items():
        if isinstance(v, float):
            print(f"{k}: {v:.4f}")

    print("\nTest metrics:")
    for k, v in test_metrics_clean.items():
        if isinstance(v, float):
            print(f"{k}: {v:.4f}")

    # --------------------------------------------------------
    # Save best model only
    # Selection criterion: validation macro-F1
    # --------------------------------------------------------

    current_val_macro_f1 = val_metrics_clean.get("val_macro_f1", -1)

    if current_val_macro_f1 > best_val_macro_f1:
        print(f"\nNew best model found: {spec['label']}")
        print(
            f"Validation macro-F1 improved from "
            f"{best_val_macro_f1:.4f} to {current_val_macro_f1:.4f}"
        )

        best_val_macro_f1 = current_val_macro_f1

        if os.path.exists(BEST_MODEL_DIR):
            shutil.rmtree(BEST_MODEL_DIR)

        os.makedirs(BEST_MODEL_DIR, exist_ok=True)

        trainer.save_model(BEST_MODEL_DIR)
        tokenizer.save_pretrained(BEST_MODEL_DIR)

        best_model_info = {
            "model_key": model_key,
            "model_label": spec["label"],
            "model_name": spec["model_name"],
            "selection_metric": "validation_macro_f1",
            "validation_macro_f1": float(current_val_macro_f1),
            "test_macro_f1": float(test_metrics_clean["test_macro_f1"]),
            "test_weighted_f1": float(test_metrics_clean["test_weighted_f1"]),
            "test_accuracy": float(test_metrics_clean["test_accuracy"]),
            "max_length": MAX_LENGTH,
            "num_labels": num_labels,
            "labels": labels,
            "label2id": label2id,
            "id2label": {str(k): v for k, v in id2label.items()}
        }

        with open(
            os.path.join(BEST_MODEL_DIR, "best_model_info.json"),
            "w",
            encoding="utf-8"
        ) as f:
            json.dump(best_model_info, f, indent=2, ensure_ascii=False)

    # --------------------------------------------------------
    # Clean memory after each model
    # --------------------------------------------------------

    del trainer
    del model
    del tokenizer
    del train_dataset
    del val_dataset
    del test_dataset
    del data_collator

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# ============================================================
# 12. Save model comparison summary
# ============================================================

comparison_df = pd.DataFrame(all_metrics)

comparison_path = os.path.join(
    OUTPUT_DIR,
    "model_comparison_summary.csv"
)

comparison_df.to_csv(comparison_path, index=False, encoding="utf-8-sig")

print("\n" + "=" * 80)
print("Model comparison summary")
print("=" * 80)

comparison_display_cols = [
    "model_label",
    "val_macro_f1",
    "val_weighted_f1",
    "val_accuracy",
    "test_macro_f1",
    "test_weighted_f1",
    "test_accuracy",
    "train_runtime_minutes"
]

display(
    comparison_df[comparison_display_cols]
    .sort_values("val_macro_f1", ascending=False)
)


# ============================================================
# 13. Save comparison figures
# ============================================================

comparison_sorted = comparison_df.sort_values(
    "val_macro_f1",
    ascending=False
).copy()

save_metric_bar_chart(
    comparison_sorted,
    metric_col="val_macro_f1",
    title="Model Comparison by Validation Macro-F1",
    save_path=os.path.join(OUTPUT_DIR, "model_comparison_validation_macro_f1.svg")
)

save_metric_bar_chart(
    comparison_sorted,
    metric_col="test_macro_f1",
    title="Model Comparison by Test Macro-F1",
    save_path=os.path.join(OUTPUT_DIR, "model_comparison_test_macro_f1.svg")
)

save_metric_bar_chart(
    comparison_sorted,
    metric_col="test_weighted_f1",
    title="Model Comparison by Test Weighted-F1",
    save_path=os.path.join(OUTPUT_DIR, "model_comparison_test_weighted_f1.svg")
)

save_metric_bar_chart(
    comparison_sorted,
    metric_col="test_accuracy",
    title="Model Comparison by Test Accuracy",
    save_path=os.path.join(OUTPUT_DIR, "model_comparison_test_accuracy.svg")
)


# ============================================================
# 14. Combine class-wise performance across models
# ============================================================

classwise_rows = []

for model_key, spec in MODEL_SPECS.items():
    report_path = os.path.join(
        OUTPUT_DIR,
        f"classification_report_{model_key}.csv"
    )

    report_df = pd.read_csv(report_path, encoding="utf-8-sig")

    temp = report_df[
        report_df["classification"].isin(labels)
    ].copy()

    temp["model_key"] = model_key
    temp["model_label"] = spec["label"]
    temp["model_label_id"] = temp["classification"].map(label2id)

    classwise_rows.append(temp)

classwise_df = pd.concat(classwise_rows, ignore_index=True)

classwise_output_path = os.path.join(
    OUTPUT_DIR,
    "classwise_performance_all_models.csv"
)

classwise_df.to_csv(
    classwise_output_path,
    index=False,
    encoding="utf-8-sig"
)

print("\nClass-wise performance:")
display(
    classwise_df[
        [
            "model_label",
            "model_label_id",
            "classification",
            "precision",
            "recall",
            "f1-score",
            "support"
        ]
    ]
)


# ============================================================
# 15. Save class-wise F1 figures
# ============================================================

for model_key, spec in MODEL_SPECS.items():
    temp = classwise_df[classwise_df["model_key"] == model_key].copy()
    temp = temp.sort_values("model_label_id").copy()

    fig, ax = plt.subplots(figsize=(11, 6))

    x = np.arange(len(temp))
    ax.bar(x, temp["f1-score"])

    ax.set_xticks(x)
    ax.set_xticklabels(
        [
            f"{int(row.model_label_id)}. {row.classification}"
            for _, row in temp.iterrows()
        ],
        rotation=45,
        ha="right"
    )

    ax.set_ylabel("F1-score")
    ax.set_title(f"Class-wise F1-score: {spec['label']}")

    for i, v in enumerate(temp["f1-score"]):
        ax.text(i, v, f"{v:.2f}", ha="center", va="bottom")

    fig.tight_layout()

    fig_path = os.path.join(
        OUTPUT_DIR,
        f"classwise_f1_{model_key}.svg"
    )

    plt.savefig(fig_path, format="svg", bbox_inches="tight")
    plt.show()
    plt.close()


# ============================================================
# 16. Save final best model summary
# ============================================================

best_summary_path = os.path.join(
    OUTPUT_DIR,
    "best_model_summary.csv"
)

best_summary_df = pd.DataFrame([best_model_info])
best_summary_df.to_csv(best_summary_path, index=False, encoding="utf-8-sig")

print("\n" + "=" * 80)
print("Best model saved")
print("=" * 80)

print("Best model directory:")
print(BEST_MODEL_DIR)

print("\nBest model info:")
print(json.dumps(best_model_info, indent=2, ensure_ascii=False))


# ============================================================
# 17. Remove temporary local files
# ============================================================

if os.path.exists(TEMP_RUNS_DIR):
    shutil.rmtree(TEMP_RUNS_DIR)

print("\nTemporary local files removed.")

print("\nAll outputs saved in:")
print(OUTPUT_DIR)
```
